##4.BUS TICKET DISCOUNT

```python
age=int(input("Enter your Age:"))
bus_ticket_price=int(input("Enter the Bus Ticket Price:"))
if age<5:
  print("Ticket price 0")
elif age>=5 and age<=18:
  Discount=bus_ticket_price *0.5##50%
  final_bus_ticket_price=bus_ticket_price-Discount
  print(" Final Ticket price discount:",final_bus_ticket_price)
elif  age>=60:
  Discount=bus_ticket_price *0.3##30%
  final_bus_ticket_price=bus_ticket_price-Discount
  print(" Final Tcket price discount:",final_bus_ticket_price)
else:
  Discount=0
  print("Give Full price")

```

##5.RESTSTAURANT BILL WITH DISCOUNT
